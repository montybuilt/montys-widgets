### Understanding Bitwise Operators

Bitwise operators in Python are a set of operators that perform operations at the bit level on integers. They manipulate individual bits within the binary representation of numbers, allowing bit-level computations. These operators enable you to work directly with the binary representations of integers, providing a way to perform operations that are not easily achievable with regular arithmetic or logical operators.

The category of bitwise operators in Python includes:

1. Bitwise AND (`&`): Performs a bitwise AND operation on the corresponding bits of two integers. It sets a bit to 1 if and only if both corresponding bits are 1.

2. Bitwise OR (`|`): Performs a bitwise OR operation on the corresponding bits of two integers. It sets a bit to 1 if at least one of the corresponding bits is 1.

3. Bitwise XOR (`^`): Performs a bitwise exclusive OR (XOR) operation on the corresponding bits of two integers. It sets a bit to 1 if the corresponding bits are different.

4. Bitwise NOT (`~`): Performs a bitwise negation (NOT) operation on a single integer. It inverts all the bits, turning 0s into 1s and 1s into 0s.

5. Left Shift (`<<`): Shifts the bits of an integer to the left by a specified number of positions, effectively multiplying the number by 2 raised to the power of the shift count.

6. Right Shift (`>>`): Shifts the bits of an integer to the right by a specified number of positions, effectively dividing the number by 2 raised to the power of the shift count (for non-negative integers).

These operators are fundamental for performing bit manipulation, setting or clearing specific bits, checking for the presence of certain flags, or optimizing specific algorithms that depend on bitwise operations.

It's important to note that while bitwise operators provide powerful low-level manipulation capabilities, they might not be as commonly used in everyday Python programming compared to regular arithmetic or logical operations, which operate at a higher level of abstraction.

### Bitwise And Operator

Bitwise AND (&) is used to evaluate bits of two integers, bit by bit - if bits at matching positions are both 1, it returns a 1 in that position.

Let's evaluate the following expression: 12 & 21

The binary representation of 12 is 1100
The binary representation of 21 is 10101

Let's line up each bit to the corresponding value position

|Number|16s|8s|4s|2s|1s|
|:---:|:---:|:---|:---:|:---:|:---:|
|12|0|1|1|0|0|
|21|1|0|1|0|1|

Now let's apply Bitwise AND to each digit:

|Number|16s|8s|4s|2s|1s|
|:---:|:---:|:---|:---:|:---:|:---:|
|12|0|1|1|0|0|
|21|1|0|1|0|1|
|Result|0|0|1|0|0|

So the resulting binary number from 12 & 21 is 100, which is a decimal equivalent of 4



```python
# lets breakdown the following expression 12 & 21:

# The binary representations of 12 and 21 are 1100 and 10101:
print(bin(12), bin(21))

# The binary representation of 12 & 21 is 100:
print(bin(12 & 21))

# The decimal representation of 100 is 4:
print(int('0b100', 2))
```

### Bitwise OR Operator

Bitwise OR (&) is used to evaluate bits of two integers, bit by bit - if either bit at matching positions is 1, it returns a 1 in that position.

Let's evaluate the following expression: 12 | 21

The binary representation of 12 is 1100
The binary representation of 21 is 10101

Let's line up each bit to the corresponding value position

|Number|16s|8s|4s|2s|1s|
|:---:|:---:|:---|:---:|:---:|:---:|
|12|0|1|1|0|0|
|21|1|0|1|0|1|

Now let's apply Bitwise OR to each digit:

|Number|16s|8s|4s|2s|1s|
|:---:|:---:|:---|:---:|:---:|:---:|
|12|0|1|1|0|0|
|21|1|0|1|0|1|
|Result|1|1|1|0|1|

So the resulting binary number from 12 & 21 is 11101, which is a decimal equivalent of 29

```python
# lets breakdown the following expression 12 | 21:

# The binary representations of 12 and 21 are 1100 and 10101:
print(bin(12), bin(21))

# The binary representation of 12 | 21 is 11101:
print(bin(12 | 21))

# The decimal representation of 100 is 29:
print(int('0b11101', 2))
```

### Bitwise XOR Operator

Bitwise XOR (`^`): Performs a bitwise exclusive OR (XOR) operation on the corresponding bits of two integers. It sets a bit to 1 if the corresponding bits are different.

Let's evaluate the following expression: 12 ^ 21

The binary representation of 12 is 1100
The binary representation of 21 is 10101

Let's line up each bit to the corresponding value position

|Number|16s|8s|4s|2s|1s|
|:---:|:---:|:---|:---:|:---:|:---:|
|12|0|1|1|0|0|
|21|1|0|1|0|1|

Now let's apply Bitwise XOR to each digit:

|Number|16s|8s|4s|2s|1s|
|:---:|:---:|:---|:---:|:---:|:---:|
|12|0|1|1|0|0|
|21|1|0|1|0|1|
|Result|1|1|0|0|1|

So the resulting binary number from 12 & 21 is 11001, which is a decimal equivalent of 25

```python
# lets breakdown the following expression 12 ^ 21:

# The binary representations of 12 and 21 are 1100 and 10101:
print(bin(12), bin(21))

# The binary representation of 12 | 21 is 11101:
print(bin(12 ^ 21))

# The decimal representation of 100 is 25:
print(int('0b11001', 2))
```

### Bitwise Shift Operators

In Python, the bitwise shift operators add or remove bits to a binary number and return the value as a decimal.  The sytax is as follows:

a << n - shifts the binary equivalent of a left by n bits and returns the decimal equivalanet

- Left shifts have the effect of adding two bits on the right
- For example 4<<2 is equivalent of shifting 100 to 10000
- This is the decimal equivalent of shifting 4 to 16

a >> n - shifts the binary equivalane of a right by n bits and returns the decimal equivalent

- Right shifts have the effect of removing the two bits on the right
- For example 4>>2 is equivalanet of shifting 100 to 1
- This is the decimal equivalent of shifting 4 to 1

```python
# Consider the decimal 17 and the bitwise left shift 17<<3

# The binary representation of 17 is 0b10001:
print(bin(17))

# The binary representation of 17<<3 is 0b10001000:
print(bin(17<<3))

# The decimal representation fo 0b10001000 is 136:
print(17<<3)

```

```python
# Consider the decimal 684 and a bitwise right shift 684>>5:

# The binary representation of 125 is 0b1010101100:
print(bin(684))

# The binary representation of 125>>6 is 0b10101:
print(bin(684>>5))

# The decimal representation of 0b10101 is 21:
print(684>>5)
```
